Load SAV File

```python
import pyreadstat
import pandas as pd

sav_path = "../data/Pew Research Center Global Attitudes Spring 2025 Dataset.sav"
_, meta = pyreadstat.read_sav(sav_path, metadataonly=True)

```

Topline data from PDF to compare with clean data

```python
topline_q30={
    "India":[14, 32, 35, 19],
    "Kenya":[12, 36, 49, 4],
    "Japan":[53, 36, 11, 0],
    "Germany":[51, 45, 4, 0],
    "Canada":[41, 51, 8, 0]

}

topline_q31={
    "India":[16, 19, 39, 26],
    "Kenya":[17, 31, 43, 10],
    "Japan":[16, 28, 55, 1],
    "Germany":[17, 29, 53, 1],
    "Canada":[9, 45, 45, 1]
}

country_code = meta.variable_value_labels["country"]

def country_to_code(name):
    for key, value in country_code.items():
        if value == name:
            return int(key)
    return None


```

```python
country_code_q30 = []
country_code_q31 = []

for name in topline_q30:
    country_code_q30.append(country_to_code(name))


for name in topline_q31:
    country_code_q31.append(country_to_code(name))
```

q_30 without weights

```python
unweight_q30 = []

for code in country_code_q30:
    name = country_code[code]
    answer = clean_df.loc[clean_df["country"] == code, "ai_heard"].value_counts()

    pct = answer / answer.sum() * 100

    unweight_q30 = [round(pct.get(1,0)),round(pct.get(2,0)),round(pct.get(3,0)),round(pct.get(8,0)+pct.get(9,0))]
    print(name,unweight_q30)
    
```

q31 without weights

```python
unweight_q31 = []

for code in country_code_q31:
    name = country_code[code]
    answer = clean_df.loc[clean_df["country"] == code, "ai_cncexc"].value_counts()

    pct = answer / answer.sum() * 100

    unweight_q31 = [round(pct.get(1,0)),round(pct.get(2,0)),round(pct.get(3,0)),round(pct.get(8,0)+pct.get(9,0))]
    

```

q_30 vertification

```python
country_code_q30 = []


for name in topline_q30:
    country_code_q30.append(country_to_code(name))

for code in country_code_q30:
    name = country_code[code]
    rows = clean_df[clean_df["country"] == code][["ai_heard", "weight"]].copy()
    answer = rows.groupby("ai_heard")["weight"].sum()

    pct = answer / answer.sum() * 100
    clean_data = [round(pct.get(1,0)),round(pct.get(2,0)),round(pct.get(3,0)),round(pct.get(8,0)+pct.get(9,0))]


    vert = True
    for i in range(len(clean_data)):
        if(abs(clean_data[i] - topline_q30[name][i]) > 1):
            vert = False


    print(name,"PDF: ",topline_q30[name],"clean data: ",vert)

```

q31 vertification

```python
country_code_q31 = []
diff = []

for name in topline_q31:
    country_code_q31.append(country_to_code(name))

for code in country_code_q31:
    name = country_code[code]
    rows = clean_df[clean_df["country"] == code]
    
    answer = rows.groupby("ai_cncexc")["weight"].sum()
    pct = answer / answer.sum() * 100
    clean_data = [round(pct.get(1,0)),round(pct.get(2,0)),round(pct.get(3,0)),round(pct.get(8,0)+pct.get(9,0))]

    vert = True
    for i in range(len(clean_data)):
        if(abs(clean_data[i] - topline_q31[name][i]) > 1):
            vert = False


    print(name,"PDF: ",topline_q31[name],"clean data: ",clean_data,vert)
```
