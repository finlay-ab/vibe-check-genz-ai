# Decicion to be made on keeping 97 as it will represent 97+ or removing it as it is to broad and will affect data. 

- 97 = 97 or older
- 98 = dont know
- 99 = refused to answer

```python
# setting 98 and 99 to NaN for age_clean

clean_df.loc[
    clean_df["age_clean"].isin([98, 99]),
    "age_clean"
] = pd.NA
```

# aus age is in age bands so we can not merge them

```python
print(
    clean_df.loc[
        clean_df["country"] == 13,
        "age_australia"
    ].value_counts(dropna=False).sort_index()
)
```

```python
# clean aus_age as 2 missing values 
clean_df["age_australia_clean"] = (
    clean_df["age_australia"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)
```

```python
australia_age_labels = {
    "1": "18-29",
    "2": "30-49",
    "3": "50-64",
    "4": "65+"
}

clean_df["age_australia_label"] = (
    clean_df["age_australia_clean"]
    .map(australia_age_labels)
)
```

## Created d_age_aus to match d_age_...

```python
# add d_age_aus

clean_df["d_age_aus"] = (
    clean_df["age_australia"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)
```

```python
australia_age_labels = {
    "1": "18-29",
    "2": "30-49",
    "3": "50-64",
    "4": "65+"
}

clean_df["d_age_aus_label"] = (
    clean_df["d_age_aus"]
    .map(australia_age_labels)
)
```

```python
australia_mask = clean_df["country"] == 13

display(
    clean_df.loc[
        australia_mask,
        [
            "age_australia",
            "d_age_aus",
            "d_age_aus_label"
        ]
    ].value_counts(dropna=False).reset_index(name="respondents")
)
```

Adding a column called 'age_range' so ages can be used with Australia's age range system

```python
age_bands = ["18-29", "30-49", "50-64", "65+"]

clean_df["age_range"] = pd.cut(
    clean_df["age_clean"],
    bins=[17, 29, 49, 64, np.inf],
    labels=age_bands
).astype("object")

# use Australia's labels
australia_mask = clean_df["country"] == 13
clean_df.loc[australia_mask, "age_range"] = clean_df.loc[australia_mask, "d_age_aus_label"]
```

```python
# Check: every country should have all four bands, and very few missing
print("Missing age_range:", clean_df["age_range"].isna().sum())

display(
    pd.crosstab(clean_df["country"], clean_df["age_range"], dropna=False)
)
```
