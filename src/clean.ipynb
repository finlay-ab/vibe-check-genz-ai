# Decicion to be made on keeping 97 as it will represent 97+ or removing it as it is to broad and will affect data. 

- 97 = 97 or older
- 98 = dont know
- 99 = refused to answer

```python
# setting 98 and 99 to NaN for age_clean

clean_df.loc[
    clean_df["age_clean"].isin([98, 99]),
    "age_clean"
] = pd.NA
```

# aus age is in age bands so we can not merge them

```python
print(
    clean_df.loc[
        clean_df["country"] == 13,
        "age_australia"
    ].value_counts(dropna=False).sort_index()
)
```

```python
# clean aus_age as 2 missing values 
clean_df["age_australia_clean"] = (
    clean_df["age_australia"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)
```

```python
australia_age_labels = {
    "1": "18-29",
    "2": "30-49",
    "3": "50-64",
    "4": "65+"
}

clean_df["age_australia_label"] = (
    clean_df["age_australia_clean"]
    .map(australia_age_labels)
)
```

## Created d_age_aus to match d_age_...

```python
# add d_age_aus

clean_df["d_age_aus"] = (
    clean_df["age_australia"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)
```

```python
australia_age_labels = {
    "1": "18-29",
    "2": "30-49",
    "3": "50-64",
    "4": "65+"
}

clean_df["d_age_aus_label"] = (
    clean_df["d_age_aus"]
    .map(australia_age_labels)
)
```

```python
australia_mask = clean_df["country"] == 13

display(
    clean_df.loc[
        australia_mask,
        [
            "age_australia",
            "d_age_aus",
            "d_age_aus_label"
        ]
    ].value_counts(dropna=False).reset_index(name="respondents")
)
```

# clean gender

```python
# merge aus gender into gender

clean_df["gender_clean"] = (
    clean_df["gender"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)

clean_df["gender_australia_clean"] = (
    clean_df["gender_australia"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)

aus_ = clean_df["country"] == 13
clean_df.loc[aus_, "gender_clean"] = clean_df.loc[aus_, "gender_australia_clean"]
```

```python
# clean missing values for q30, q31 and q32
```

```python
# setting 8, 9, 98, 99 to NaN for q30 and q31
# change col names to the actual variable names from the codebook

q30_q31_cols = ['q30_col_name', 'q31_col_name'] 

for col in q30_q31_cols:
    if col in clean_df.columns:
        clean_df.loc[
            clean_df[col].isin([8, 9, 98, 99, '8', '9', '98', '99']),
            col
        ] = pd.NA
```

```python
# setting 9, 98, 99 to NaN for q32 

q32_cols = ['q32a_col_name', 'q32b_col_name', 'q32c_col_name', 'q32d_col_name']

for col in q32_cols:
    if col in clean_df.columns:
        clean_df.loc[
            clean_df[col].isin([9, 98, 99, '9', '98', '99']),
            col
        ] = pd.NA
```

```python
age_bands = ["18-29", "30-49", "50-64", "65+"]

clean_df["age_range"] = pd.cut(
    clean_df["age_clean"],
    bins=[17, 29, 49, 64, np.inf],
    labels=age_bands
).astype("object")

# use Australia's labels
australia_mask = clean_df["country"] == 13
clean_df.loc[australia_mask, "age_range"] = clean_df.loc[australia_mask, "d_age_aus_label"]
```
